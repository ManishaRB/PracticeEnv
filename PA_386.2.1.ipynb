# Data Type Objects in NumPy

```python
import numpy as np
```

```python
#Create a NumPy array with dtype int32
arr_int = np.array([1, 2, 3, 4], dtype='int32')
print(arr_int)
print(arr_int.dtype)
```

```python
#Create a NumPy array with dtype float64
arr_float = np.array([1.5, 2.5, 3.5], dtype='float64')
print(arr_float)
print(arr_float.dtype)
```

```python
#One‑Dimensional Array
#Create a 1D array from 1 to 5
arr = np.array([1, 2, 3, 4, 5])
print(arr)
```

```python
arr = np.arange(1, 6)
print(arr)
```

```python
#NumPy Operations
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

print("Addition:", arr1 + arr2)
print("Subtraction:", arr1 - arr2)
print("Multiplication:", arr1 * arr2)
print("Division:", arr1 / arr2)
```

```python
#Dot product
dot_product = np.dot(arr1, arr2)
print("Dot Product:", dot_product)
```

```python
#Mean, Median, Standard Deviation
print("Mean:", np.mean(arr1))
print("Median:", np.median(arr1))
print("Standard Deviation:", np.std(arr1))
```
