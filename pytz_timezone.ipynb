# Time Zone Conversion w/Pytz

```python
# Timezone conversion with Pytz
# install pytx: pip install pytz
import pytz
from datetime import datetime

# create a function with pytz to convert timezones

# Accepts 3 parameters:
# dt - the date/time
# from_tz - for current tz
# to_tz - timezone to convert time too
def convert_time_zone(dt, from_tz, to_tz):
    
    # create timezone objects
    # taking tz string and creating special tz object with pytz
    from_tz = pytz.timezone(from_tz)
    to_tz = pytz.timezone(to_tz)

    # Convert/add the tz to the datetime object if necessary
    # setting the tz attribute
    dt = dt.replace(tzinfo=from_tz)

    # converting datetime to new timezone
    dt = dt.astimezone(to_tz)

    return dt

# ---------------------------------------------------
# Using created function
today_datetime = datetime.now()

from_timezone = 'US/Eastern'
to_timezone = 'Asia/Tokyo'

converted_time = convert_time_zone(today_datetime, from_timezone, to_timezone)

print(converted_time)
```
